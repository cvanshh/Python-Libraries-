```python
import numpy as np
```

```python
# View: Creates a new view of the original array without copying data.
# Changes made to the view can affect the original array.

# Copy: Creates an independent copy of the original array.
# Changes made to the copy do not affect the original array.
```

```python
#view

arr = np.array([1,2,3,4,5,6])

view = arr[1:3]
view[0]=99

print(arr)  #array got affected
```

```python
#making a copy arr

arr = np.array([1,2,3,4,5,6])
copy_arr = arr.copy()

copy_arr[0] = 99

print(arr)

print(copy_arr)
```

```python
# check wether two arrays are sharig the same memory

from numpy import *

arr = np.array([1, 2, 3, 4, 5])
view = arr[1:3]
copy = arr.copy()

print(may_share_memory(arr, view)) 
print(may_share_memory(arr, copy))
```

```python
# Shallow Copy: Creates a new object but keeps references to the nested objects.
# Changes to nested objects in the copy can affect the original object.

# Deep Copy: Creates a completely independent copy, including nested objects.
# Changes made to the copy do not affect the original object.
```
