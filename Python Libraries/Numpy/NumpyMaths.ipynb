```python
import numpy as np
```

```python
#Basic Arihmetic Ops

arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([2, 4, 5, 8, 10])

# Addition: Adds corresponding elements of two arrays.
print("Addition:", np.add(arr1, arr2))

# Subtraction: Subtracts corresponding elements of two arrays.
print("Subtraction:", np.subtract(arr1, arr2))

# Multiplication: Multiplies corresponding elements of two arrays.
print("Multiplication:", np.multiply(arr1, arr2))

# Division: Divides corresponding elements of two arrays.
print("Division:", np.divide(arr1, arr2))

# Floor Division: Returns the rounded-down quotient after division.
print("Floor Division:", np.floor_divide(arr1, arr2))

# Modulus: Returns the remainder after division.
print("Modulus:", np.mod(arr1, arr2))

# Power: Raises each element to the power of the corresponding element.
print("Power:", np.power(arr1, arr2))

# Reciprocal: Returns the multiplicative inverse (1/x) of each element.
print("Reciprocal:", np.reciprocal(arr2.astype(float)))
```

```python
#Aggregate Functions

arr = np.array([10, 20, 30, 40, 50])

# Sum: Returns the sum of all array elements.
print("Sum:", np.sum(arr))

# Mean: Returns the average of array elements.
print("Mean:", np.mean(arr))

# Median: Returns the middle value of sorted array elements.
print("Median:", np.median(arr))

# Minimum: Returns the smallest element.
print("Minimum:", np.min(arr))

# Maximum: Returns the largest element.
print("Maximum:", np.max(arr))

# Standard Deviation: Measures how spread out values are from the mean.
print("Standard Deviation:", np.std(arr))

# Variance: Measures the average squared deviation from the mean.
print("Variance:", np.var(arr))

# Cumulative Sum: Returns the running sum of array elements.
print("Cumulative Sum:", np.cumsum(arr))

# Cumulative Product: Returns the running product of array elements.
print("Cumulative Product:", np.cumprod(arr))

# Product: Returns the product of all array elements.
print("Product:", np.prod(arr))
```

```python
#Exp and Log
arr = np.array([1, 2, 3, 4, 5])

# Square Root: Returns the square root of each element.
print("Square Root:", np.sqrt(arr))

# Square: Returns the square of each element.
print("Square:", np.square(arr))

# Exponential: Calculates e raised to the power of each element.
print("Exponential:", np.exp(arr))

# Natural Log: Calculates logarithm with base e.
print("Natural Log:", np.log(arr))

# Log10: Calculates logarithm with base 10.
print("Log10:", np.log10(arr))

# Log2: Calculates logarithm with base 2.
print("Log2:", np.log2(arr))
```

```python
# Rounding Functionsarr = np.array([10.25, 20.75, 30.45, 40.85, 50.15])

# Round: Rounds values to the nearest specified decimal place.
print("Round:", np.round(arr, 1))

# Floor: Rounds values down to the nearest integer.
print("Floor:", np.floor(arr))

# Ceiling: Rounds values up to the nearest integer.
print("Ceiling:", np.ceil(arr))

# Truncate: Removes the decimal part of each value.
print("Truncate:", np.trunc(arr))
```

```python
#Other OPs
arr = np.array([-10, -20, 0, 30, 40])

# Absolute: Returns the non-negative value of each element.
print("Absolute:", np.abs(arr))

# Sign: Returns -1 for negative, 0 for zero and 1 for positive values.
print("Sign:", np.sign(arr))

# Dot Product: Multiplies corresponding elements and returns their sum.
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Dot Product:", np.dot(a, b))
```

```python
# Trigo Ops
import numpy as np

# Angles in degrees
arr = np.array([0, 30, 45, 60, 90])

# Convert degrees into radians (NumPy trigonometric functions use radians).
rad = np.radians(arr)

# Sin: Calculates the sine of each angle.
print("Sin:", np.sin(rad))

# Cos: Calculates the cosine of each angle.
print("Cos:", np.cos(rad))

# Tan: Calculates the tangent of each angle.
print("Tan:", np.tan(rad))

# Arcsin: Calculates the inverse sine and returns angles in radians.
print("Arcsin:", np.arcsin([0, 0.5, 1]))

# Arccos: Calculates the inverse cosine and returns angles in radians.
print("Arccos:", np.arccos([0, 0.5, 1]))

# Arctan: Calculates the inverse tangent and returns angles in radians.
print("Arctan:", np.arctan([0, 0.5, 1]))
```
