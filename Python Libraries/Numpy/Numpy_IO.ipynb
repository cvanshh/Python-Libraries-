```python
import numpy as np
```

```python
data = np.array([[85, 90], [75, 80], [95, 92]])
np.savetxt("scores.csv", data, delimiter=",", fmt="%d")
```

```python
loaded = np.loadtxt("scores.csv" , delimiter=',')
print(loaded)
```

```python
arr = np.array([1, 2, 3, 4])
np.save("my_array.npy", arr)

loaded_arr = np.load("my_array.npy")
print(loaded_arr)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
np.savez("multiple_arrays.npz", array1=a, array2=b)

data = np.load("multiple_arrays.npz")
print(data["array1"])  # [1 2 3]
print(data["array2"])  # [4 5 6]
```

```python
# .npy (NumPy Array): Stores a single NumPy array in binary format.

# .npz (NumPy Compressed Archive): Stores multiple NumPy arrays in one file.

# .csv (Comma-Separated Values): Stores tabular data separated by commas.

# .txt (Text File): Stores data in a simple, human-readable text format.
```
