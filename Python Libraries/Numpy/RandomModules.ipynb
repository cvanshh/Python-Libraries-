```python
import numpy as np
```

```python
# NumPy Random: A module used to generate random numbers and perform random sampling and Testing Data.
```

```python
# np.random.rand(): Generates random values between 0 and 1.
np.random.seed(
    42
)  # np.random.seed(): Sets the random seed to produce reproducible results.
print(np.random.rand((3)))  # 1D array
print(np.random.rand(2, 2))  # 2D array


# We use a random seed to get consistent results, compare ML models fairly,
# debug errors easily, and reproduce experiments.
```

```python
# np.random.randint(): Generates random integers within a specified range.

print(np.random.randint(10 ,20, size=5))
#(from 10, to 20, 5 numbers)
```

```python
# np.random.randn(): Generates random numbers from a standard normal distribution.
# mean = 0, std = 1

print(np.random.randn(4))

print(np.random.randn(4,4))
```

```python
# np.random.random(): Generates random floating-point numbers between 0 and 1.
print(np.random.random(5))
```

```python
items = ["Apple", "Banana", "Orange"]

# np.random.choice(): Selects random elements from an array.
print(np.random.choice(items, size=2))

# np.random.shuffle(): Randomly shuffles the elements of an array (inplace).
np.random.shuffle(items)
print(items)
```

```python
#permutation : working is same like shuffel but create a new array

arr = np.array([10,20,30,40])

permutated = np.random.permutation(arr)

print("Original: ", arr)
print("permutated: ", permutated)
```
