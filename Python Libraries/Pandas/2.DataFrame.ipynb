```python
import numpy as np
import pandas as pd 
```

```python
# DataFrame: A 2D labeled data structure in Pandas, consisting of rows and columns.
# It can store different data types and is commonly used for data manipulation and analysis.
# DataFrame: Basically Collection of Series
```

```python
data = {
    "Name": ["John", "Anna", "Peter", "Linda"],
    "Age": [28, 34, 29, 42],
    "City": ["New York", "Paris", "Berlin", "London"],
    "Salary": [65000, 70000, 62000, 85000],
}
df = pd.DataFrame(data)

print(df)
```

```python
# DataFrame using list
data_list = [
    ["John", 28, "New York", 65000],
    ["Anna", 34, "Paris", 70000],
    ["Peter", 29, "Berlin", 62000],
    ["Linda", 42, "London", 85000],
]
df2 = pd.DataFrame(data_list)
print(df2)  #since we dint provide labels it came like this 
```

```python
data_list = [
    ["John", 28, "New York", 65000],
    ["Anna", 34, "Paris", 70000],
    ["Peter", 29, "Berlin", 62000],
    ["Linda", 42, "London", 85000],
]
col_list = ["Name", "Age", "City", "Salary"]
index_list = [1,2,3,4]
df2 = pd.DataFrame(data_list,index=index_list,columns= col_list)
print(df2)
```

```python
#Selection and indexing of columns
```

```python
#df2[0]  will not work as this is series not the index
df2["Name"]
```

```python
# if we need to print multiple cols we just need to pass the list inside the function

df2[["Name", "City"]]
```

```python
#Creating a new col

df2["Designation"] = ["Doctor","Engineer","Doctor","Engineer"]
df2
```

```python
#Remove a col
df2.drop("Designation",axis = 1,inplace= True) #inplace make sure its deleted permanently from RAM 
                                       #axis = 0 for rows (default)
                                       #axis = 1 for cols (we need to change Manually)
```

```python
#Selecting multiple rows
df2.loc[[1,2]]
```

```python
#Selecting Subsets of rows and cols

df.loc[[1,2]][["City","Salary"]]
```

```python
#Conditioinal Selections
#I only want to see those people whose age is above 30

df2[df2["Age"]>30]
```
