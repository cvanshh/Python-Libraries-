```python
import pandas as pd 
import numpy as np
```

```python
#Find Missing Data

data = {
    'A': [1, 2, np.nan, 4, 5],           #Nan is Null value(Not A Number)
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)

print(df)
```

```python
show_null_values = df.isna()         #when Nan then True else False
count_null_values = df.isna().sum()  #count number of null values per col

print(show_null_values)
print()
print(count_null_values)
```

```python
df.isna().any()  #check if all cols having null values
```

```python
#Removing Missing Values(Work on the basis of Rows)

df.dropna()   #sirf vo rows bachengi jisme koi bhi null values nhi hai
```

```python
df.dropna(thresh = 3)  #thresh=k : atlease k not null values
```

Filling the Missing Data


```python
df.fillna(0)
```

```python
values  = {'A':0 , 'B': 100, 'C':300,  'D':400} 
df.fillna(value=values)     #Assigning Values what we want
```

```python
#df.isna()   : Show Full DF with Missing Values
#df.dropna() : Drop Missing Values Row-Wise
#df.fillna() : Fill Missing values Col-Wise
```

```python
df.fillna(df.mean())   #Handle Missing Values with Mean of DF
```
