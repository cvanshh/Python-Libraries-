```python
import numpy as np
import pandas as pd
```

GrouBy

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Group by Category and calculate the sum of Sales
cat = df.groupby('Category')  #already done at backend
```

```python
#to show what made at backend
for i,v in cat:
    print(i)
    print(v)
```

```python
#now sum of category
sales_sum_by_cat = df.groupby('Category')['Sales'].sum()

print(sales_sum_by_cat)
```

```python
# Group by Store and calculate the sum of Sales
df.groupby('Store')['Sales'].sum()
```

```python
# Group by multiple columns
# Group by Category and Store

df.groupby(['Category','Store'])
```
