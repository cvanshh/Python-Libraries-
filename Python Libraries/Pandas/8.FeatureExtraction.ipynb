```python
import numpy as np
import pandas as pd
```

```python
 df = pd.read_csv(r'anime.csv')
 #r stands for raw strings : It tells Python to treat backslashes (\) as literal characters instead of escape characters
```

```python
df.head()
```

```python
#Make a new col of episode count
df.loc[2]['Title']
```

```python
# Number of episodes extraction
def extract_episodes(txt):
    data = ""
    check = False
    for ch in txt:
        if ch == "(":
            check = True
        if check == True:
            data += ch
        if ch == ")":
            check = False
    return data

    
```

```python
df['No. of Episodes'] = df['Title'].apply(extract_episodes)
```

```python
df.head(2)
```

```python
df["No. of Episodes"] = df["No. of Episodes"].str.replace("eps", "", regex=False)
```

```python
df.head()
```

```python
# Remove parentheses and extra spaces
df["No. of Episodes"] = (
    df["No. of Episodes"].str.replace(r"[()]", "", regex=True).str.strip()
)

# Convert into integer datatype
df["No. of Episodes"] = df["No. of Episodes"].astype(int)
```

```python
df.head()
```

```python
#Make a new column for time stamp
df.loc[1]['Title']
```

```python
def extraction_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                data += txt[j]
    return data
```

```python
df['TimeStamp'] = df['Title'].apply(extraction_time)
```

```python
df.head(2)
```

```python
#Extracting Months [Code Copied From Gpt]

from dateutil.relativedelta import relativedelta
from datetime import datetime


def calculate_total_months(period):
    try:
        start_str, end_str = period.split(" - ")
        start_date = datetime.strptime(start_str, "%b %Y")
        end_date = datetime.strptime(end_str, "%b %Y")
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1  # +1 to include the starting month
    except:
        return None


df["Months"] = df["TimeStamp"].apply(calculate_total_months)
```
